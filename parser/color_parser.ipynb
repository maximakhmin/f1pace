```python
from sqlalchemy import create_engine, text
import os
from dotenv import load_dotenv

engine = create_engine(f'postgresql://{os.getenv('DB_USER')}:{os.getenv('DB_PASSWORD')}@localhost:5432/f1pace')
```

```python
import pandas as pd
```

```python
sessions = pd.read_sql("SELECT s.id, st.name, e.year, e.round FROM sessions s JOIN events e ON s.event_id = e.id JOIN session_types st ON s.session_type=st.id", engine)
sessions
```

```python
from fastf1 import set_log_level

set_log_level("WARNING")
```

```python
from fastf1 import get_session
from fastf1.plotting import list_driver_abbreviations, get_driver_style, get_team_name_by_driver
from tqdm import tqdm

style_list = ['linestyle', 'marker', 'color', 'facecolor', 'edgecolor']
columns = ['session_id', 'driver_abbr', 'team'] + style_list

for i in tqdm(range(len(sessions))):

    with engine.connect() as conn:
        cursor = conn.execute(text(f"SELECT * FROM style_info WHERE session_id = {sessions.loc[i, 'id']}"))
        curr_s = cursor.fetchall()
    if len(curr_s)!=0:
        print(f"Уже есть запись session_id={sessions.loc[i, 'id']}")
        continue

    try:

        data = []

        fastf1_session = get_session(sessions.loc[i, "year"],
                                    sessions.loc[i, "round"],
                                    sessions.loc[i, "name"])
        
        drivers_abbrs = list_driver_abbreviations(fastf1_session)
        for driver_abbr in drivers_abbrs:
            style = get_driver_style(driver_abbr, style_list, fastf1_session, exact_match=True)
            data.append([
                sessions.loc[i, "id"],
                driver_abbr,
                get_team_name_by_driver(driver_abbr, fastf1_session, short=True, exact_match=True),
                ] +
                [style[attr] for attr in style_list]
            )

        pd.DataFrame(data, columns=columns).to_sql('style_info', engine, index=False, if_exists='append')
    
    except Exception as e:
        print(e)
        print(f"Ошибка session_id={sessions.loc[i, 'id']}")
```
