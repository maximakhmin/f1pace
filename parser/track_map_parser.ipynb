```python
from sqlalchemy import create_engine, text
import os
from dotenv import load_dotenv

engine = create_engine(f'postgresql://{os.getenv('DB_USER')}:{os.getenv('DB_PASSWORD')}@localhost:5432/f1pace')
```

```python
with engine.connect() as con:
    cursor = con.execute(text("SELECT * FROM events"))
    events = cursor.fetchall()

events
```

```python
from fastf1 import get_session, set_log_level
from tqdm import tqdm

set_log_level("ERROR")

for event in tqdm(events):

    event_id = event[0]
    event_year = event[2]
    event_round = event[3]

    with engine.connect() as con:
        cursor = con.execute(text(f"SELECT * FROM track_corners WHERE event_id = {event_id}"))
        track_corners = cursor.fetchall()
    if len(track_corners)!=0:
        print("{event} already exist")
        continue
    
    try:
        session = get_session(event_year, event_round, 1)
        session.load()
        corners = session.get_circuit_info().corners[["X", "Y", "Angle", "Number"]]
        corners.columns = ["x", 'y', 'angle', 'number']
        corners["event_id"] = event_id
        corners.to_sql('track_corners', engine, if_exists='append', index=False)
    except Exception as e:
        print(event, e)
```
